```python
import spacy
from spacy import displacy
from spacy import tokenizer
import re

nlp = spacy.load("en_core_web_sm")
txt = "google was founded in 1974 and computer scientist larry page and sergey brin while they found good reason"
spacy_doc = nlp(txt)

for word in spacy_doc.ents:
    print(word.text, word.label_)
```

```python

displacy.render(spacy_doc, style='ent', jupyter=False)
```
